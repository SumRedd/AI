# Do gamblers have a higher chance of speeding?

### I will be conducting an observational study because I have access to a group of people rather than a population.

### My Analysis

```python
import pandas as pd
df = pd.read_csv("steak.csv")
df = df.drop(0)
df.groupby("Do you ever gamble?")["Do you ever drive above the speed limit?"].value_counts(normalize=True)
```

```python
df.groupby("Do you ever gamble?")["Do you ever drive above the speed limit?"].apply(
    lambda x: (x == "Yes").mean()
).plot(kind="bar")
```

### My Conclusion
My data indicates that gammblers have a higher relative frequency to speed than non-gamblers. 92.6% of gamblers said they speed, compared to about 86.0% of non-gamblers.

# What is the average tempo of SZA songs in Spotify's 2023 dataset?

#### I will be conducting an observational study because I am analyzing tempo data for songs already affiliated with SZA on Spotify's 2023 dataset. I will not be manipulating variables. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("spotify-2023.csv", encoding="latin-1")
```

```python
sza_songs = df[df["artist(s)_name"].str.contains("SZA", case=False, na=False)]

bpm = sza_songs["bpm"]
```

```python
average_bpm = sza_songs["bpm"].mean()
median_bpm = sza_songs["bpm"].median()

print("Mean:", average_bpm)
print("Median:", median_bpm)
```

```python
sns.histplot(data=sza_songs, x="bpm")

plt.axvline(average_bpm, linestyle="--", label=f"Mean: {average_bpm:.1f}")
plt.axvline(median_bpm, linestyle="--", label=f"Median: {median_bpm:.1f}")

plt.xlabel("Tempo (BPM)")
plt.ylabel("Number of Songs")
plt.title("Distribution of SZA Song Tempos")
plt.legend()
plt.show()
```

## The average tempo of SZA songs in Spotify's 2023 dataset is **117.0 BPM** according to the mean value.

### The histogram above shows the distribution of SZA songs across BPM ranges 65-160, with a mean line at 117 BPM, and a median line at 119. I chose this graph to emphasize SZA's variety in BPM across the 2023 Spotify dataset.

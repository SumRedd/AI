### My Question
What is the probability of rolling seven on a dice with twenty sides if rolled once.

### My Study
Describe your simulation here:
1. We will simulate a fair twenty-sided dice being rolled once.
2. One trial will be rolling the dice once. The possible outputs from that trial are integers 1-20.
3. Rolling a seven is the event of interest.

```python
import numpy as np
num_rolls = 1000
rolls = np.random.randint(1, 21, size=num_rolls)
```

```python
count_sevens = np.sum(rolls == 7)
simulated_probability = np.mean(rolls == 7)
print(f"Number of times 7 was rolled: {count_sevens}")
print(f"Estimated (Simulated) Probability: {simulated_probability:.4f}")
```

```python
favorable_outcomes = 1
total_outcomes = 20
theoretical_probability = favorable_outcomes / total_outcomes
print(f"Theoretical Probability: {theoretical_probability:.4f}")
```

### My Conclusion
The theoretical probability of a twenty sided dice rolling a seven is 0.0500 (5%). This was calculated as 1/20. The empircal probability of a twenty sided dice rolling a seven is 0.0520 (5.20%). There were 52 counts of seven rolled when the dice was rolled 1,000 times. The simulation results align with my expectations because of how similar theoretical probabilty and empirical probability were.

```python
import matplotlib.pyplot as plt

outcomes, counts = np.unique(rolls, return_counts=True)
relative_frequencies = counts / len(rolls)

plt.figure(figsize=(10, 5))
plt.bar(outcomes, relative_frequencies, color='skyblue', edgecolor='black', label='Empirical Probability')
plt.axhline(y=1/20, color='red', linestyle='--', linewidth=2, label='Theoretical Probability (0.05)')

plt.xlabel('Die Outcome (1-20)')
plt.ylabel('Relative Frequency')
plt.title('Simulated vs. Theoretical Probability for a 20-Sided Die (1000 Rolls)')
plt.xticks(range(1, 21))
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.7)
plt.show()
```

The barchart encapsulates our empirical probability in relation to theoretical probability (red-dashed line, 0.05).
It supports our answer because the empirical probability of 7 sits just above our theoretical probability, thus showing mathematical allignment.

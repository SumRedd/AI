# What is the relative frequency of poisonous mushrooms that grow on dung?

## I will be conducting an observational study because I am observing a portion of the mushroom population and its traits.

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("mushrooms.csv")
```

```python
dung = df[df["habitat"] == "d"]
```

```python
relative_frequency = (dung["class"] == "p").mean()
print(relative_frequency)
```

# The relative frequency of poisonous mushrooms that grow on dung is 40.28%.

```python
sns.countplot(data=dung, x="class")
```

### The graph shows there are more edible mushrooms than poisonous mushrooms that grow on dung.
#### 1,880 are edible.
#### 1,270 are poisonous.

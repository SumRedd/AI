### **How do the z-scores for Colorado School of Mines and CU Boulder compare when evaluating their low-income graduation rates against 4-year-institution national averages?**

#### I will be conducting a census study because I have access to the entire population of colleges in the U.S., which has been kept up to date by the U.S. Department of Education. I will not be manipulating variables.

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("cleaned_college.csv")
df4yr = df[df["school_level_code"] == 1]

national_gr = df4yr["grad_rate_4yr_school"].mean()
national_std = df4yr["grad_rate_4yr_school"].std()

mines_pellgr = df4yr.loc[
    df4yr["school_name"] == "Colorado School of Mines",
    "grad_rate_pell_4yr_school",
].values[0]
boulder_pellgr = df4yr.loc[
    df4yr["school_name"] == "University of Colorado Boulder",
    "grad_rate_pell_4yr_school",
].values[0]

mines_z = (mines_pellgr - national_gr) / national_std
boulder_z = (boulder_pellgr - national_gr) / national_std

schools = ["CU Boulder", "Colorado School of Mines"]
z_scores = [boulder_z, mines_z]

plt.figure(figsize=(8, 3.5))
bars = plt.barh(schools, z_scores, color=["#CFB87C", "#8B0000"], height=0.5)
plt.axvline(0, color="gray", linestyle="--", label="National Average (z = 0)")

plt.bar_label(bars, fmt="z = %.2f", padding=5, fontweight="bold")

plt.xlim(0, 1.2)
plt.xlabel("Z-Score (Standard Deviations Above Average)")
plt.title("Low-Income Graduation Rate Z-Scores")
plt.legend(loc="lower right")
plt.tight_layout()

plt.show()
```

#### I chose this graph to highlight how Pell graduation rates for both schools are slightly higher than the national average (z=0).

### **The z-scores for Colorado School of Mines and CU Boulder, regarding Pell graduation rates, fall slightly higher than the U.S. national average. Colorado School of Mines is almost one standard deviation above average, at 0.91, and CU Boulder is also slightly higher at 0.69.**
